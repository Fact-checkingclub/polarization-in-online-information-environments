```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from statsmodels.graphics.tsaplots import plot_acf
```

```python

community_a = pd.read_csv('./result/weekly/all features/fill_nan/version2/askscience1.csv')
community_a['community'] = 'Askscience'

community_b = pd.read_csv('./result/weekly/all features/fill_nan/version2/conspiracy1.csv')
community_b['community'] = 'Conspiracy'


data = pd.concat([community_a, community_b], axis=0)


print(data.head())


data.to_csv('./result/weekly/all features/fill_nan/version2/merged_community_data_science_conspiracy.csv', index=False)  

```

```python

sns.set(style="white")  


community_colors = {
    'Askscience': '#A80326',
    'Conspiracy': '#469EB4'
}


palette = sns.color_palette("Set1", n_colors=2)


plt.figure(figsize=(4, 2))


for i, community in enumerate(data['community'].unique()):
    
    community_data = data[data['community'] == community]
    
    
    sns.kdeplot(community_data['entropy_x_filled'], shade=True, label=community, color=community_colors[community])


plt.legend(title="", fontsize=9, loc='upper right', markerscale=0.1)


plt.title('KDE of entropy-based linguistic complexity')
plt.xlabel('')
plt.ylabel('Density')


plt.show()
```

```python
import numpy as np
from scipy.stats import mannwhitneyu

def compute_effect_size_r(u_stat, n1, n2):
    mean_U = n1 * n2 / 2
    std_U = np.sqrt(n1 * n2 * (n1 + n2 + 1) / 12)
    z = (u_stat - mean_U) / std_U
    r = abs(z) / np.sqrt(n1 + n2)
    return r



community1 = pd.read_csv('./result/weekly/all features/fill_nan/version2/askscience1.csv')
community2 = pd.read_csv('./result/weekly/all features/fill_nan/version2/conspiracy1.csv')


features = ["entropy_x_filled","Gini_filled","taaco_filled","density_filled","entropy_y_filled","similarity_filled","clustering_filled","transitivity_filled","distance_filled",
            "betweenness_centrality_filled",
                "in-group identity_filled","out-group identity_filled","gender identity_filled","queer identity_filled","disability identity_filled","body size and physical identity_filled","relational identity_filled",
                "professional identity_filled","partisan identity_filled",
                "generational identity_filled","ideological identity_filled","racial identity_filled","class identity_filled","religious identity_filled","national identity_filled",
                "anger_filled","anticipation_filled","disgust_filled","fear_filled","joy_filled","sadness_filled","surprise_filled","trust_filled"]  

results = []

for feature in features:
    x = community1[feature].dropna()
    y = community2[feature].dropna()
    
    n1, n2 = len(x), len(y)
    
    u_stat, pval = mannwhitneyu(x, y, alternative='two-sided')
    r = compute_effect_size_r(u_stat, n1, n2)
    
    results.append({
        'feature': feature,
        'U': u_stat,
        'p-value': pval,
        'effect_size_r': r,
        'n1': n1,
        'n2': n2
    })


results_df = pd.DataFrame(results)
print(results_df.sort_values(by='p-value'))
```

```python
from statsmodels.tsa.stattools import adfuller, kpss
import pandas as pd

def stationarity_test(series, name="", alpha=0.05):
    print(f"\n====== 平稳性检验：{name} ======")

    # Drop NA
    series = series.dropna()

    # ---- ADF ----
    adf_result = adfuller(series, autolag='AIC')
    adf_stat, adf_p = adf_result[0], adf_result[1]
    print(f"ADF: statistic={adf_stat:.4f}, p={adf_p:.4f}")
    if adf_p < alpha:
        print(" → ADF 结论：拒绝单位根假设 → 序列平稳")
        adf_flag = True
    else:
        print(" → ADF 结论：无法拒绝单位根假设 → 序列非平稳")
        adf_flag = False

    # ---- KPSS ----
    try:
        kpss_result = kpss(series, regression='ct')
        kpss_stat, kpss_p = kpss_result[0], kpss_result[1]
        print(f"KPSS: statistic={kpss_stat:.4f}, p={kpss_p:.4f}")
        if kpss_p < alpha:
            print(" → KPSS 结论：拒绝平稳假设 → 序列非平稳")
            kpss_flag = False
        else:
            print(" → KPSS 结论：无法拒绝平稳假设 → 序列平稳")
            kpss_flag = True
    except:
        print("KPSS 无法运行（序列过短或常数列）")
        kpss_flag = False

    
    print("\n📌 综合判断：", end="")
    if adf_flag and kpss_flag:
        print("序列显著平稳 ✔")
    elif (not adf_flag) and (not kpss_flag):
        print("序列显著非平稳，需要差分")
    else:
        print("结果不一致：可能为趋势平稳 → 建议差分或加入趋势项")

    print("=====================================\n")


# -------------------------

# -------------------------


cols_to_test = ["Polar_filled","Home_dist_filled","Home_link_filled"]  


for col in cols_to_test:
    stationarity_test(df[col], name=col)
```
